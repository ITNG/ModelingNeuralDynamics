# Chapter 35: Periodic inhibition

Representative deterministic normalized LIF comparison with $I=0.2$, $α=5$, period 25 ms, mean inhibitory conductance 0.1, and 100 ms observation, matching PERIODIC_INHIBITION. The normalized gate has unit mean before scaling. Tonic inhibition uses the same mean. Noise, conductance-based RTM and f–I sweeps are omitted.

Reference: [Python chapter](../python/chapter35.ipynb) and [chapter guide](../python/chapter35.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.drives import simulate_periodic_lif, periodic_shape
```

```
periodic = simulate_periodic_lif(.2)
tonic = simulate_periodic_lif(.2, periodic=False)
print("Periodic / tonic rates (Hz):", periodic.rate, tonic.rate)
```

```
fig, axes = plt.subplots(2, 1, sharex=True)
axes[0].plot(periodic.t, .1*periodic_shape(periodic.t)); axes[0].axhline(.1, color='r'); axes[0].set_ylabel('Conductance')
for r, label in [(periodic,'Periodic'),(tonic,'Tonic')]:
    axes[1].plot(r.t,r.v,label=label)
axes[1].set(xlabel='Time (ms)',ylabel='Normalized voltage'); axes[1].legend()
fig.tight_layout(); fig.savefig(output/'chapter35-inhibition.png')
```
