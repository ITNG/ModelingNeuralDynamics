# Chapter 37: Thresholding in PING

Representative single-cell thresholding experiment uses the source periodic inhibitory waveform $\exp(\cos^4(\pi t/25))-1$, normalized to mean conductance 0.15. Seventeen independent RTM cells sample drives 0–3 for the full source 5000 ms observation. This discrete recruitment curve illustrates thresholding without claiming the source bisection-derived window widths. Full PING networks, the never-reset reference and boundary bisection are omitted.

Reference: [Python chapter](../python/chapter37.ipynb) and [chapter guide](../python/chapter37.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.drives import simulate_rtm_forcing
```

```
currents = np.linspace(0., 3., 17)
r = simulate_rtm_forcing(currents, kind="inhibition", power=4, alpha=1., g_bar=.15, duration=5000.)
print("Sampled rates (Hz):", r.rates)
```

```
fig, ax = plt.subplots()
ax.plot(currents,r.rates,'o-'); ax.set(xlabel='Current (uA)',ylabel='Response rate (Hz)')
fig.tight_layout(); fig.savefig(output/'chapter37-recruitment.png')
```
