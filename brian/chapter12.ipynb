# Chapter 12: Two-dimensional bifurcation analysis — Brian2

Representative simulations adapted from [`python/chapter12.ipynb`](../python/chapter12.ipynb) and its [chapter notes](../python/chapter12.md).

Coverage: the reduced RTM invariant cycle at I=1, using the reference trajectory's initial state. The long fixed-point/eigenvalue continuation and near-saddle zoom remain in the Python reference; this notebook illustrates the dynamical part rather than claiming a bifurcation diagram.

The RTM reduction imposes $m=m_\infty(v)$ and $h=1-n$, leaving voltage and potassium activation as the two dynamic states.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_reduced
out=simulate_reduced("rtm",current=1,initial=(-75.58244579620455,.005935897840905),duration=30,dt=.005)
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].plot(out.t,out.v[0]);ax[0].set(xlabel="t (ms)",ylabel="v (mV)")
ax[1].plot(out.v[0],out.n[0]);ax[1].set(xlabel="v (mV)",ylabel="n")
plt.tight_layout();plt.show()
assert np.max(out.v)>0
```
