# Chapter 38: Gamma coherence

Representative GAMMA_COHERENCE_1 two-cell RTM/WB circuit: a sharp 25 ms main drive and broad 40 ms distractor each have mean 0.5. E→I conductance 0.35, I→E 0.5, I→I 0.15, full 200 ms duration. Timing markers allow inspection of inhibitory response alignment. The initial voltage has deterministic small heterogeneity rather than the exact source resting voltage. Mean-inhibition comparisons and large Poisson-PING phase sweeps are omitted.

Reference: [Python chapter](../python/chapter38.ipynb) and [chapter guide](../python/chapter38.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.networks import simulate_two_cell_coherence
```

```
r = simulate_two_cell_coherence(duration=200.)
```

```
fig, ax = plt.subplots(figsize=(9, 4))
ax.scatter(r.e_spikes.t/b2.ms, r.e_spikes.i, s=5, label='E')
ax.scatter(r.i_spikes.t/b2.ms, r.i_spikes.i+len(r.e), s=5, label='I')
ax.set(xlabel='Time (ms)', ylabel='Cell index')
ax.legend()
fig.tight_layout()
for t in np.arange(0,201,25): ax.axvline(t,color='gray',alpha=.2)
fig.savefig(output/'chapter38-coherence.png')
fig, ax = plt.subplots()
ax.plot(r.i_state.t/b2.ms, r.i_state.s_ie[0],label='Inhibitory gate')
ax.set(xlabel='Time (ms)',ylabel='Gate'); ax.legend()
fig.savefig(output/'chapter38-inhibition.png')
```
