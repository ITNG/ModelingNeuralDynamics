# Chapter 39: Short-term depression and facilitation

Representative RTM depression (200 ms) and depression plus facilitation (400 ms), using the source dynamic sodium activation m. Continuous voltage release $\gamma=1.545(1+	anh(v/10))$ transfers available pool p into q; recovery is $(1-p-q)/	au_{rec}$. Gate s is driven by q. Facilitation evolves log-utilization W using the source exponential recovery. These are presynaptic release demonstrations, without added postsynaptic feedback. The pulse-area illustration and WB variant are omitted.

Reference: [Python chapter](../python/chapter39.ipynb) and [chapter guide](../python/chapter39.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.plasticity import simulate_short_term
```

```
depression = simulate_short_term(duration=200.)
facilitation = simulate_short_term(duration=400., facilitation=True)
assert np.all(facilitation.p+facilitation.q <= 1.00001)
```

```
fig, axes = plt.subplots(3, 1, sharex=True, figsize=(9,7))
for r,label in [(depression,'Depression'),(facilitation,'Depression + facilitation')]:
    axes[0].plot(r.t,r.v,label=label); axes[1].plot(r.t,r.s,label=label); axes[2].plot(r.t,r.p,label=label)
axes[0].set_ylabel('v (mV)'); axes[1].set_ylabel('Gate s'); axes[2].set(xlabel='Time (ms)',ylabel='Available p')
axes[0].legend(); fig.tight_layout(); fig.savefig(output/'chapter39-plasticity.png')
fig, ax = plt.subplots(); ax.plot(facilitation.t,facilitation.U)
ax.set(xlabel='Time (ms)',ylabel='Utilization 1-exp(-W)'); fig.savefig(output/'chapter39-utilization.png')
```
