# Chapter 32: Weak PING rhythms

Representative M-current PING: 40 E and 10 I cells, 1000 ms, $g_M=1$, source mean drives 3 and 0.7. This deliberately smaller all-to-all network exposes the adaptation mechanism at modest execution cost. Unlike the source 200/50 example, it retains source-scale 5% Gaussian drive heterogeneity but starts near rest without phase-splay initialization; its raster is not a figure reproduction. Poisson variants, phase maps and cluster sweeps are omitted. The reusable network helper also supports independent one-to-one Poisson inputs.

Reference: [Python chapter](../python/chapter32.ipynb) and [chapter guide](../python/chapter32.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.networks import simulate_ping
```

```
rng = np.random.default_rng(63806)
current_e = 3.*(1+.05*rng.standard_normal(40))
current_i = .7*(1+.05*rng.standard_normal(10))
r = simulate_ping(n_e=40, n_i=10, current_e=current_e, current_i=current_i, g_m=1., duration=1000.)
assert np.all(np.isfinite(r.e_state.v))
```

```
fig, ax = plt.subplots(figsize=(9, 4))
ax.scatter(r.e_spikes.t/b2.ms, r.e_spikes.i, s=5, label='E')
ax.scatter(r.i_spikes.t/b2.ms, r.i_spikes.i+len(r.e), s=5, label='I')
ax.set(xlabel='Time (ms)', ylabel='Cell index')
ax.legend()
fig.tight_layout()
fig.savefig(output/'chapter32-raster.png')
fig, axes = plt.subplots(2, 1, sharex=True)
axes[0].plot(r.e_state.t/b2.ms, r.e_state.v[0]); axes[0].set_ylabel('v (mV)')
axes[1].plot(r.e_state.t/b2.ms, r.e_state.w[0]); axes[1].set(xlabel='Time (ms)',ylabel='M gate w')
fig.tight_layout(); fig.savefig(output/'chapter32-adaptation.png')
```
