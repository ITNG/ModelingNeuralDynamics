# Chapter 16: Model neurons of bifurcation type 3 — Brian2

Representative simulations adapted from [`python/chapter16.ipynb`](../python/chapter16.ipynb) and its [chapter notes](../python/chapter16.md).

Coverage: the discrete-feedback self-exciting theta neuron (source defaults). Crossing pi wraps theta to -pi and SETS the feedback to w_max. INaP–IK fixed-point/phase-plane and smooth-feedback counterparts remain in the Python reference. This example isolates the feedback mechanism; it does not equate every self-exciting trajectory with tonic type-3 firing.

$\dot\theta=1-\cos\theta+(I+w)(1+\cos\theta)$, $\dot w=-w/20$, with $I=-0.05$ and spike reset $w=0.2$.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_self_exciting_theta
out=simulate_self_exciting_theta()
fig,ax=plt.subplots(2,1,figsize=(10,5),sharex=True)
ax[0].plot(out.t,out.theta[0]);ax[0].set_ylabel("theta")
ax[1].plot(out.t,out.w[0]);ax[1].set(xlabel="model time",ylabel="feedback w")
for t in out.spike_t:
    ax[0].axvline(t,color="grey",alpha=.3)
plt.tight_layout();plt.show()
assert len(out.spike_t)>0 and np.max(out.w)<=.2+1e-12
print("Spikes:",len(out.spike_t))
```
