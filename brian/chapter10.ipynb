# Chapter 10: The slow–fast phase plane — Brian2

Representative simulations adapted from [`python/chapter10.ipynb`](../python/chapter10.ipynb) and its [chapter notes](../python/chapter10.md).

Coverage: FHN nullclines and orbit, followed by a reduced HH trajectory. Fixed-point continuation and the cycle-speed coloring panels remain in the Python reference. FHN uses its source initial state (-1,-2); RK4 permits a 0.05 model-time step here. Reduced HH retains the source initial state (-50,0.4).

FHN: $\dot v=v-v^3/3-n+I$, $\dot n=(av-n)/\tau_n$. Reduced HH imposes $m=m_\infty(v)$ and $h=0.83-n$.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_fhn, simulate_reduced
fhn = simulate_fhn(initial=(-1,-2), duration=400, dt=.05)
vgrid = np.linspace(-2.5,2.5,400)
fig, ax = plt.subplots(1,2,figsize=(11,4))
ax[0].plot(vgrid,vgrid-vgrid**3/3-.5,label="v nullcline")
ax[0].plot(vgrid,1.25*vgrid,label="n nullcline")
ax[0].plot(fhn.v[0],fhn.n[0],color="k",label="Brian2 orbit")
ax[0].set(xlabel="v",ylabel="n");ax[0].legend()
ax[1].plot(fhn.t,fhn.v[0]);ax[1].set(xlabel="model time",ylabel="v")
plt.tight_layout();plt.show()
assert np.ptp(fhn.v[0,-4000:]) > 2
```

```
hh = simulate_reduced("hh", current=10, initial=(-50,.4), duration=50, dt=.002)
fig, ax = plt.subplots(1,2,figsize=(10,4))
ax[0].plot(hh.t,hh.v[0]);ax[0].set(xlabel="t (ms)",ylabel="v (mV)")
ax[1].plot(hh.v[0],hh.n[0]);ax[1].set(xlabel="v (mV)",ylabel="n")
plt.tight_layout();plt.show()
assert np.isfinite(hh.v).all()
```
