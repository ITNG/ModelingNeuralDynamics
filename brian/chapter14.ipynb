Representative simulations adapted from [`python/chapter14.ipynb`](../python/chapter14.ipynb) and its [chapter notes](../python/chapter14.md).

Coverage: reduced HH attracting trajectories at the reference current I=5.5 and at I=10. The reduced Erisir comparison, fixed-point/eigenvalue scans, and backward-traced repelling cycles remain in the Python reference. These forward trajectories illustrate the cycle geometry; they do not establish the precise Hopf current.

Reduced HH uses $h=0.83-n$ and instantaneous sodium activation. The HH conductances are gNa=120, gK=36, gL=0.3, with reversal potentials 45,-82,-59 mV.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_reduced
fig,axes=plt.subplots(1,2,figsize=(10,4))
for current in [5.5,10]:
    out=simulate_reduced("hh",current=current,initial=(-50,.4),duration=150,dt=.002)
    tail=out.t>50
    axes[0].plot(out.t[tail],out.v[0,tail],label=f"I={current}")
    axes[1].plot(out.v[0,tail],out.n[0,tail],label=f"I={current}")
axes[0].set(xlabel="t (ms)",ylabel="v (mV)");axes[0].legend()
axes[1].set(xlabel="v (mV)",ylabel="n");axes[1].legend()
plt.tight_layout();plt.show()
assert np.max(out.v)>0
```
