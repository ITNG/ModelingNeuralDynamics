# Chapter 27: Phase locking with delays

Actual delayed Brian synaptic events reproduce the chapter's two abstract pulse
oscillators. Compare delays of 0.1 and 0.7 natural periods using the source PRC
g(phi)=phi(1-phi)/3, initial phases (0,0.9), and the initial A spike at time zero.
One natural period is one millisecond of Brian time.

Reference: [Python chapter 27](../python/chapter27.ipynb). Coverage: delayed
phase-oscillator pairs. Three-cell cases and the theta parameter grid remain in
Python. In particular, this example does not reuse the Python theta-pair helper's
acknowledged update bug: each phase follows its own ODE and queued arrivals.

Run from a source checkout after `uv sync --all-groups`, with this checkout's
`.venv` kernel. The new shared helpers are not assumed to be on PyPI.


```
from pathlib import Path
import sys
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "mnd").is_dir())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
import numpy as np
import matplotlib.pyplot as plt
from mnd.brian.phases import theta_prc, phase_pair, weak_difference, lif_inhibition
```

```
results = [phase_pair(kind='delayed', epsilon=1/3, initial_phases=(0., .9),
                      initial_spike=True, delay=delay, duration=20.)
           for delay in (.1, .7)]
fig, axes = plt.subplots(2, 1, figsize=(10, 4.5), sharex=True)
for ax, result in zip(axes, results):
    assert np.isfinite(result.phase).all()
    ax.scatter(result.spike_t, result.spike_i, marker='|', s=130)
    ax.set(yticks=[0, 1], yticklabels=['A', 'B'], title=f'delay = {result.delay} periods',
           ylim=(-.4, 1.4))
axes[-1].set_xlabel('time / intrinsic period')
plt.tight_layout()
plt.show()
```
