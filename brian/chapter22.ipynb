# Chapter 22: A Wilson–Cowan oscillatory E–I network — Brian2

Representative simulations adapted from [`python/chapter22.ipynb`](../python/chapter22.ipynb) and its [chapter notes](../python/chapter22.md).

Coverage: the original rectified/saturating population-rate ODE, plus reduced recurrent excitation. Nullcline construction and synthetic rate-sampled raster remain in the Python reference. These are population activities, not individual-neuron voltages; using Brian2 for the rate equations does not change that interpretation.

$\tau_E\dot E=-E+f(w_{EE}E-w_{IE}I+P_E)$ and $\tau_I\dot I=-I+g(w_{EI}E-w_{II}I+P_I)$, where $f(x)=100x^2/(900+x^2)$, $g(x)=100x^2/(400+x^2)$ for x>0 and zero otherwise.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_wilson_cowan
baseline=simulate_wilson_cowan()
reduced=simulate_wilson_cowan(w_ee=1.,duration=300)
fig,ax=plt.subplots(1,2,figsize=(11,4))
ax[0].plot(baseline.t,baseline.E[0],label="E, wEE=1.5")
ax[0].plot(baseline.t,baseline.I[0],label="I, wEE=1.5")
ax[0].plot(reduced.t,reduced.E[0],"--",label="E, wEE=1.0")
ax[0].set(xlabel="t (ms)",ylabel="population activity");ax[0].legend()
ax[1].plot(baseline.E[0],baseline.I[0]);ax[1].set(xlabel="E",ylabel="I")
plt.tight_layout();plt.show()
assert np.ptp(baseline.E[0,baseline.t>150])>10
assert np.min(baseline.E)>=0 and np.max(baseline.E)<=100
```
