# Chapter 11: The saddle-node bifurcation — Brian2

Representative simulations adapted from [`python/chapter11.ipynb`](../python/chapter11.ipynb) and its [chapter notes](../python/chapter11.md).

Coverage: Brian2 trajectories before, at, and after the positive-x saddle–node collision. The equilibrium at (0,0) persists in all three cases; the original figure excludes it from view. Algebraic equilibrium markers are calculated directly. Detailed boundary-arrow placement remains in the reference.

$\dot x=-ax+y$, $\dot y=x^2/(1+x^2)-y$, with $a=0.45,0.5,0.55$ and $b=1$.

Run `uv sync --group dev` in the repository root and select that environment's Python kernel. These examples require the source checkout containing `mnd/brian`; an older PyPI installation does not contain these new helpers. Run all cells in order. Brian2 performs every trajectory integration. Numerical states use the book's units; helper clocks map a source time unit to one Brian millisecond. FHN and normal-form time is abstract model time.

```
from pathlib import Path
import sys

# Use this source checkout so newly added helpers are available.
root = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "mnd").is_dir()), None)
if root is not None:
    sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
b2.prefs.codegen.target = "numpy"
```

```
from mnd.brian.dynamics import simulate_saddle_node
fig, axes = plt.subplots(1,3,figsize=(12,4))
for ax,a in zip(axes,[.45,.5,.55]):
    for initial in [(.3,.1),(.7,.2),(1.5,.3),(1.8,.8)]:
        out=simulate_saddle_node(a=a,initial=initial,duration=100,dt=.05)
        ax.plot(out.x[0],out.y[0],lw=1)
    if a <= .5:
        roots=np.unique([(1-np.sqrt(1-4*a*a))/(2*a),(1+np.sqrt(1-4*a*a))/(2*a)])
        ax.plot(roots,a*roots,"ko",label="positive equilibria")
    ax.set(xlim=(.2,2),ylim=(0,1),xlabel="x",title=f"a={a}")
axes[0].set_ylabel("y")
plt.tight_layout();plt.show()
```
