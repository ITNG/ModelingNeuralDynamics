# Chapter 28: Weakly coupled oscillators

The weak-coupling reduction is itself an ODE that Brian can integrate:
psi' = epsilon [g(psi)-g(-psi)], where g(phi) = p²(1-p), p=phi mod 1.
Both terms are evaluated periodically. Zeros and their restoring flow determine
phase locking. Compare trajectories on slow time epsilon*t.

Reference: [Python chapter 28](../python/chapter28.ipynb),
`simulate_weakly_coupled_de` and `wc1_g`. Coverage: the identical-pair reduced ODE
for two coupling strengths and two initial phases; analytical roots,
heterogeneous pairs, and event-versus-reduction overlays remain in Python.
One source time unit is represented by one millisecond in Brian.

Run from a source checkout after `uv sync --all-groups`, with this checkout's
`.venv` kernel. The new shared helpers are not assumed to be on PyPI.


```
from pathlib import Path
import sys
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "mnd").is_dir())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
import numpy as np
import matplotlib.pyplot as plt
from mnd.brian.phases import theta_prc, phase_pair, weak_difference, lif_inhibition
```

```
fig, ax = plt.subplots(figsize=(7, 4))
results = []
for epsilon in (.1, .5):
    for initial_phase in (.1, .4):
        result = weak_difference(epsilon=epsilon, initial_phase=initial_phase)
        results.append(result)
        assert np.isfinite(result.psi).all()
        ax.plot(epsilon*result.t, result.psi,
                label=f'epsilon={epsilon}, initial={initial_phase}')
ax.set(xlabel='slow time: epsilon × t', ylabel='phase difference')
ax.legend()
plt.tight_layout()
plt.show()
```
