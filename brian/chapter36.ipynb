# Chapter 36: f–I curves with pulsed excitation

Representative RTM comparison retains the source 1000 ms observation and 25 ms smooth pulse period ($α=1$, leak 0.1), while explicitly choosing a compact 13-point mean-current grid from 0 to 2 instead of the source 201 points. Pulsed and tonic currents have equal mean. Rates count all upward crossings during the observation; no transient removal. Both conditions start independently at resting gates. This differs from the source constant-drive continuation and first-interspike-interval estimator; the pulsed source also continues state between drive values. The second leak value and idealized schematics are omitted.

Reference: [Python chapter](../python/chapter36.ipynb) and [chapter guide](../python/chapter36.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.drives import simulate_rtm_forcing
```

```
currents = np.linspace(0., 2., 13)
constant = simulate_rtm_forcing(currents, kind="constant", duration=1000.)
pulsed = simulate_rtm_forcing(currents, kind="pulsed", duration=1000.)
```

```
fig, ax = plt.subplots()
ax.plot(currents,constant.rates,'o-',label='Constant'); ax.plot(currents,pulsed.rates,'s-',label='Pulsed')
ax.set(xlabel='Mean current (uA)',ylabel='Rate (Hz)'); ax.legend()
fig.tight_layout(); fig.savefig(output/'chapter36-fi.png')
np.savez(output/'chapter36-fi.npz', currents=currents,constant=constant.rates,pulsed=pulsed.rates)
```
