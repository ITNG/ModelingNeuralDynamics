# Chapter 40: Continuous voltage-driven STDP

Representative THREE_CELL_PING_5: two RTM E cells (currents 0.4/0.8) and one WB I cell, full 500 ms duration. The adaptation trace obeys $\dot a=1-5a(1+	anh(v/10))$. E→E weights follow the source continuous voltage-triggered rule, including exponential traces, near-zero taper and soft lower/upper bounds; this is not event-pair STDP. Initial reciprocal weights 0.05 have bound 0.4. A fixed-weight control distinguishes coevolution from baseline coupling. Abbott–Song plots, coupling sweeps and the large plastic PING population are omitted.

Reference: [Python chapter](../python/chapter40.ipynb) and [chapter guide](../python/chapter40.md). The Brian2 implementation uses explicit clocks and continuous ODEs; it does not execute the reference notebook. Numerical voltage/current values use the source chapter conventions (mV/uA), and time is in ms.

Run from a local source checkout with dependencies installed (`uv sync`). These examples do not install an unreleased package from PyPI.

```
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "mnd").is_dir():
    root = root.parent
if not (root / "mnd").is_dir():
    raise RuntimeError("Run from this repository or its brian directory; install the checkout with uv sync first.")
sys.path.insert(0, str(root))
import brian2 as b2
import numpy as np
import matplotlib.pyplot as plt
output = root / "results" / "brian-expansion"
output.mkdir(parents=True, exist_ok=True)
from mnd.brian.plasticity import simulate_three_cell_stdp
```

```
fixed = simulate_three_cell_stdp(duration=500., plastic=False)
r = simulate_three_cell_stdp(duration=500.)
assert np.max(np.abs(r.weights[:,-1]-.05)) > .001
```

```
fig, ax = plt.subplots(figsize=(9, 4))
ax.scatter(r.e_spikes.t/b2.ms, r.e_spikes.i, s=5, label='E')
ax.scatter(r.i_spikes.t/b2.ms, r.i_spikes.i+len(r.e), s=5, label='I')
ax.set(xlabel='Time (ms)', ylabel='Cell index')
ax.legend()
fig.tight_layout()
fig.savefig(output/'chapter40-raster.png')
fig, ax = plt.subplots()
for j in range(2): ax.plot(r.weight_state.t/b2.ms,r.weights[j],label=f'Plastic projection {j+1}')
ax.axhline(.05,linestyle='--',color='gray',label='Fixed-weight control')
ax.set(xlabel='Time (ms)',ylabel='E to E conductance'); ax.legend()
fig.tight_layout(); fig.savefig(output/'chapter40-weights.png')
```
