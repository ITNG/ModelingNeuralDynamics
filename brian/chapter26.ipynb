# Chapter 26: Phase locking of two oscillators

Brian threshold/reset events and reciprocal synapses implement the two pulse
oscillators from [Python chapter 26](../python/chapter26.ipynb). Between pulses,
phase advances uniformly. At an arrival the receiving phase changes by g(phi).
One natural period is represented by one millisecond in Brian's clock; this is
an abstract oscillator time scale, not a biophysical prediction.

Coverage: g(phi)=phi²(1-phi) and g(phi)=2phi(1-phi)³, both from the source.
The initial phases are the source's phases after its initial A event; that event
is not reapplied. Static f/F/G map panels and RTM PRCs remain in Python.

Run from a source checkout after `uv sync --all-groups`, with this checkout's
`.venv` kernel. The new shared helpers are not assumed to be on PyPI.


```
from pathlib import Path
import sys
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "mnd").is_dir())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
import numpy as np
import matplotlib.pyplot as plt
from mnd.brian.phases import theta_prc, phase_pair, weak_difference, lif_inhibition
```

```
results = [phase_pair(kind=kind, duration=12.) for kind in ('quadratic', 'cubic')]
fig, axes = plt.subplots(2, 1, figsize=(9, 4.5), sharex=True)
for ax, result, label in zip(axes, results, ['phi²(1-phi)', '2phi(1-phi)³']):
    assert np.isfinite(result.phase).all()
    assert all(np.count_nonzero(result.spike_i == cell) > 5 for cell in range(2))
    ax.scatter(result.spike_t, result.spike_i, marker='|', s=160)
    ax.set(yticks=[0, 1], yticklabels=['A', 'B'], title=label, ylim=(-.4, 1.4))
axes[-1].set_xlabel('time / intrinsic period')
plt.tight_layout()
plt.show()
```
