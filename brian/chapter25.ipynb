# Chapter 25: Phase response curves

Measure the theta neuron's response to a voltage kick using Brian2, then compare
with the closed-form return map in [Python chapter 25](../python/chapter25.ipynb).
The phase coordinate is time divided by the unperturbed period, not a uniformly
spaced theta angle. A voltage kick therefore changes theta nonlinearly.

Coverage: theta-neuron voltage-kick PRCs at two strengths. The chapter's RTM/WB/HH/
Erisir synaptic-pulse and voltage-kick comparisons remain in the Python notebook.
Time is in milliseconds; the QIF voltage coordinate is dimensionless.

Run from a source checkout after `uv sync --all-groups`, with this checkout's
`.venv` kernel. The new shared helpers are not assumed to be on PyPI.


```
from pathlib import Path
import sys
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "mnd").is_dir())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
import numpy as np
import matplotlib.pyplot as plt
from mnd.brian.phases import theta_prc, phase_pair, weak_difference, lif_inhibition
```

```
strong = theta_prc(kick=.1)
weak = theta_prc(kick=.05)
for result in (strong, weak):
    assert np.isfinite(result.prc).all()
    np.testing.assert_allclose(result.prc, result.predicted_prc, atol=1e-4)
fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].plot(strong.phi, strong.predicted_prc, 'k-', label='closed form')
ax[0].plot(strong.phi, strong.prc, 'o', ms=3, label='Brian2')
ax[0].set(xlabel='phase', ylabel='phase advance', title='Voltage kick = 0.1')
ax[0].legend()
ax[1].plot(strong.phi, strong.prc/.1, label='kick = 0.1')
ax[1].plot(weak.phi, weak.prc/.05, label='kick = 0.05')
ax[1].set(xlabel='phase', ylabel='advance / kick', title='Approach to weak response')
ax[1].legend()
plt.tight_layout()
plt.show()
```
